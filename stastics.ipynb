### What is the difference between population and sample standard deviation?
- Population standard deviation is calculated using the formula that divides by N (the total number of data points), while sample standard deviation divides by (N-1) to account for the fact that it is an estimate of the population standard deviation based on a sample. The sample standard deviation is typically larger than the population standard deviation because it includes an additional degree of freedom to account for the variability in the sample.


## What is an outlier
- Outliers are data points that differ significantly from the rest of a dataset, appearing as unusually high or low values that do not follow the general pattern. They represent rare events, measurement errors, or natural variability, often skewing statistical results like mean and variance, making their detection crucial for accurate analysis.

### Key Aspects of Outliers
- Definition: Observations that lie an abnormal distance from other values.
- Causes: Common sources include human error (typos), measurement/instrument errors, or genuine, albeit rare, variability.
### Types:
- Global Outliers: Data points that differ significantly from the rest of the entire dataset.
- Contextual Outliers: Observations that are considered unusual only within a specific context.
- Collective Outliers: A set of data points that together are anomalous, even if they aren't individually.

* Impact: They can heavily distort statistical models, reduce model accuracy, and lead to misleading conclusions.

### Detection Methods
* Visual Techniques: Scatter plots (identifying points outside clusters), box plots (displaying points beyond whiskers), and histograms.
* Statistical Techniques:
- Interquartile Range (IQR): Data points falling below 𝑄1−1.5×IQR or above 𝑄3+1.5×IQR are considered outliers.
- Z-score: Measures how many standard deviations a point is from the mean.
- Machine Learning: Techniques such as One-Class SVM.

* Handaling Outliers
-  Outliers should not always be removed. If they represent genuine variability or important anomalies (e.g., fraud detection), they should be kept. If they are confirmed as errors or noise, they can be removed or corrected to improve analysis
